**Q1. Which branch has the highest total_sales? Which has the lowest?**

```python
highest_total_sales=df1.groupby("branch")["total_sales"].sum()
highest_total_sales
```

```python
highest_total_sales.idxmax()
```

```python
lowest_total_sales=df1.groupby("branch")["total_sales"].sum()
lowest_total_sales
```

```python
lowest_total_sales.idxmin()
```

```python
branch_sales = df1.groupby("branch")["total_sales"].sum()

highest_lowest = branch_sales.loc[
    [branch_sales.idxmax(), branch_sales.idxmin()]]

print(highest_lowest)
```

```python
# plotting bar graph based on highest_total_sales and lowest_total_sales
import matplotlib.pyplot as plt

highest_lowest.plot(kind="bar",legend="total_sales",color="yellow",figsize=(6,4))

plt.title("Branches with Highest and Lowest Total Sales")
plt.xlabel("Branch")
plt.ylabel("total_Sales")
plt.xticks(rotation=0)
plt.legend()
plt.show()
```

**Q3. How does profit trend over the 12 months of 2024**

```python
# profit trends of the 12 months of 2024
# checking whether date column is in the correct date format
df1["date"] = pd.to_datetime(df1["date"], errors="coerce")
```

```python
df1.info()
```

```python
# filtering transactions to 2024
df1["date"] = pd.to_datetime(df1["date"], errors="coerce")
```

```python
# creating variable df_2024
df_2024 = df1[df1["date"].dt.year == 2024]
```

```python
# calculating total profit for each month
monthly_profit=df_2024.groupby(df_2024["date"].dt.month)["profit"].sum()
print(monthly_profit)
```

```python
# plotting a line chart using
import matplotlib.pyplot as plt

monthly_profit.plot(kind="line", marker="o", figsize=(10, 5),color="blue",label="trends")

plt.title("Monthly Profit Trend for 2024")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
